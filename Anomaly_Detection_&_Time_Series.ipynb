# 1.  What is Anomaly Detection? Explain its types (point, contextual, and collective anomalies) with examples.

- What is Anomaly Detection?

Anomaly Detection is the process of identifying data points or patterns that significantly differ from normal behavior.

Anomalies are also called outliers. They may represent fraud, errors, equipment failures, unusual customer behavior, or security threats.

Types of Anomalies

There are three common types:

| Type                   | Meaning                                                      | Example                                                                                      |
| ---------------------- | ------------------------------------------------------------ | -------------------------------------------------------------------------------------------- |
| **Point Anomaly**      | A single data point is significantly different from the rest | A credit-card transaction of ₹5,00,000 when the customer's normal purchases are below ₹5,000 |
| **Contextual Anomaly** | A data point is unusual only in a particular context         | 35°C is normal in summer but unusual during winter                                           |
| **Collective Anomaly** | A group of related data points is unusual as a whole         | A sudden sequence of hundreds of login attempts from one account                             |


1. Point Anomaly

A point anomaly occurs when one individual observation is substantially different from the other observations.

Example:
Suppose an e-commerce customer usually spends ₹500–₹2,000 per order, but suddenly makes a purchase worth ₹2,00,000.

That individual transaction could be flagged as an anomaly.

Applications:

- Credit-card fraud detection
- Detecting abnormal sensor readings
- Identifying unusual financial transactions

2. Contextual Anomaly

A contextual anomaly is an observation that is unusual given its context, even though the value itself may not normally be unusual.

Example:
A temperature of 35°C may be normal in Delhi during summer but unusual during a winter night.

Here, the context could include:

- Location
- Time of year
- Time of day
- Weather conditions

Another example is website traffic: 10,000 visitors per hour might be normal during a major sale but abnormal at 3 AM.

3. Collective Anomaly

A collective anomaly occurs when a group or sequence of observations, considered together, behaves unusually.

Individual observations may appear normal, but their combined pattern is abnormal.

Example:
A server normally receives 5–10 login attempts per minute. Suddenly, it receives:

10 → 12 → 15 → 20 → 25 → 30 → 40 login attempts/minute

Each value might not be extreme by itself, but the overall pattern could indicate a brute-force attack.

Applications:

- Cybersecurity
- Network monitoring
- Sensor monitoring
- Time-series analysis

Simple way to remember

- Point: One unusual point
- Contextual: Unusual in a particular situation
- Collective: A group of points is unusual together

In short: Anomaly detection helps organizations identify unusual behavior that may indicate fraud, errors, failures, attacks, or other important events.


# 2. Compare Isolation Forest, DBSCAN, and Local Outlier Factor in terms of their approach and suitable use cases.

-  Isolation Forest vs DBSCAN vs Local Outlier Factor

All three methods can be used for anomaly/outlier detection, but they identify anomalies in different ways.

| Feature                        | Isolation Forest                                     | DBSCAN                                                   | Local Outlier Factor (LOF)                                 |
| ------------------------------ | ---------------------------------------------------- | -------------------------------------------------------- | ---------------------------------------------------------- |
| **Basic approach**             | Isolates unusual observations using random trees     | Finds dense regions and labels sparse points as noise    | Compares the local density of a point with its neighbors   |
| **Main idea**                  | Anomalies are easier to isolate                      | Outliers are points outside dense clusters               | Outliers have lower local density than their neighbors     |
| **Supervision**                | Unsupervised                                         | Unsupervised                                             | Unsupervised                                               |
| **Works well with**            | Large, high-dimensional datasets                     | Spatial/geometric data with irregular clusters           | Data with varying local densities                          |
| **Number of clusters needed?** | No                                                   | No                                                       | No                                                         |
| **Handles noise?**             | Yes                                                  | Yes, explicitly labels noise                             | Yes                                                        |
| **Scalability**                | Generally very good                                  | Can become expensive with large/high-dimensional data    | Can become expensive for large datasets                    |
| **Main parameters**            | `n_estimators`, `contamination`                      | `eps`, `min_samples`                                     | `n_neighbors`, `contamination`                             |
| **Typical use cases**          | Fraud detection, cybersecurity, unusual transactions | Spatial data, customer groups, detecting isolated points | Fraud detection, network anomalies, unusual local behavior |


1. Isolation Forest

Approach:
Isolation Forest randomly partitions the data using decision trees. Anomalies tend to be isolated from other observations with fewer splits.

Example:
In a banking dataset, most transactions might be between ₹500 and ₹10,000, while unusual transactions can be isolated quickly.

Suitable for:

- Credit-card fraud detection
- Large transaction datasets
- High-dimensional data
- Network intrusion detection

Key advantage: It is generally fast and scalable for large datasets.

2. DBSCAN

Approach:

DBSCAN (Density-Based Spatial Clustering of Applications with Noise) groups points based on density. Points that do not belong to sufficiently dense regions are classified as noise (-1).

Example:
Suppose customer purchasing data forms several dense groups. A customer whose behavior does not belong to any dense group may be identified as an outlier.

Suitable for:

- Geographical/spatial data
- Data with irregularly shaped clusters
- Detecting noise in clustered datasets
- Applications where both clusters and outliers are needed

Key advantage: It can find arbitrarily shaped clusters and identify noise without specifying the number of clusters.

3. Local Outlier Factor (LOF)

Approach:

LOF compares the local density of a point with the density of its neighboring points.

If a point is located in a much less dense region than its neighbors, it receives a high outlier score.

Example:
Imagine two customer groups: one has very frequent purchases and another has infrequent purchases. A customer who behaves unusually compared with customers in their immediate neighborhood can be detected by LOF.

Suitable for:

- Data where local behavior matters
- Datasets containing regions with different densities
- Fraud detection
- Detecting unusual customer behavior

Key advantage: It can identify observations that are unusual relative to their local neighborhood, rather than only globally.

Simple comparison

Isolation Forest

       ↓

"How quickly can I isolate this point?"

       ↓

Good for large/high-dimensional data


DBSCAN

       ↓

"Does this point belong to a dense region?"

       ↓

Good for spatial/irregularly shaped clusters


LOF

       ↓

"Is this point much less dense than its neighbors?"

       ↓

Good for local anomalies and varying densities

Which one should you use?
- Large, high-dimensional dataset → Isolation Forest
- Spatial data + irregular clusters + noise → DBSCAN
- Local anomalies and varying densities → LOF

In short: Isolation Forest focuses on isolation, DBSCAN focuses on density-based clustering, and LOF focuses on local density differences.


# 3.  What are the key components of a Time Series? Explain each with one example?

-  Key Components of a Time Series

A time series is a sequence of observations recorded in time order, such as daily sales, monthly revenue, or yearly temperature.

The four main components are:

| Component                         | Meaning                                                    | Example                                                                |
| --------------------------------- | ---------------------------------------------------------- | ---------------------------------------------------------------------- |
| **1. Trend**                      | Long-term upward or downward movement                      | A company's sales increasing every year                                |
| **2. Seasonality**                | A pattern that repeats at fixed and known time intervals   | Ice-cream sales increasing every summer                                |
| **3. Cyclical Variation**         | Long-term fluctuations related to economic/business cycles | Car sales falling during a recession and rising during economic growth |
| **4. Irregular/Random Variation** | Unpredictable fluctuations caused by unusual events        | Sales suddenly dropping because of a natural disaster                  |


1. Trend

A trend represents the long-term direction of the time series.

Example:
An e-commerce company's annual sales:

2022 → ₹10 lakh

2023 → ₹12 lakh

2024 → ₹15 lakh

2025 → ₹18 lakh

The overall movement is upward, so there is an increasing trend.

2. Seasonality

Seasonality refers to patterns that repeat at regular intervals, such as daily, weekly, monthly, or yearly.

Example:
A clothing store may experience higher sales every December because of holiday shopping.

December → High sales

January   → Lower sales


December → High sales

January   → Lower sales

The repeated pattern is seasonality.

3. Cyclical Variation

Cyclical variation represents longer-term rises and falls associated with business or economic cycles.

Unlike seasonality, cycles do not necessarily occur at fixed intervals.

Example:
Car sales may:

Economic growth → Sales increase

Recession       → Sales decrease

Recovery        → Sales increase

This is a cyclical pattern.

4. Irregular/Random Variation

Irregular variation consists of unpredictable changes that cannot be explained by regular trends or seasonal patterns.

Example:
A store normally has stable sales, but sales suddenly fall because of a flood, strike, pandemic, or unexpected system failure.

These unexpected fluctuations are called irregular or random variation.


Easy way to remember

Trend       → Long-term direction

Seasonality → Repeating pattern

Cyclical    → Long-term business/economic cycle

Irregular   → Unpredictable/random changes

In short:

A time series can be thought of as Trend + Seasonality + Cyclical effects + Irregular variation. Understanding these components helps analysts identify patterns and build better forecasting models.


# 4. Define Stationary in time series. How can you test and transform a non-stationary series into a stationary one?

- Stationarity in Time Series

A stationary time series is a series whose statistical properties remain constant over time.

In a stationary series:

- Mean remains approximately constant.
- Variance remains approximately constant.
- Autocorrelation depends mainly on the time lag, not on the particular time period.
- There is no systematic trend or changing seasonal pattern.
Example

A series such as:

10, 11, 9, 10, 12, 8, 11, 10

may be approximately stationary because it fluctuates around a fairly constant mean.

Whereas:

10, 15, 20, 25, 30, 35

is non-stationary because it has an increasing trend.

How to Test for Stationarity
1. Visual Inspection

Plot the time series and look for:

- Trend
- Changing variance
- Seasonality
- Increasing/decreasing fluctuations

A time-series plot is a useful first check.

2. Augmented Dickey-Fuller (ADF) Test

The ADF test is commonly used to test for a unit root.

- Null hypothesis (H₀): Series is non-stationary.
- Alternative hypothesis (H₁): Series is stationary.

Interpretation:

| p-value            | Conclusion                                |
| ------------------ | ----------------------------------------- |
| **p-value < 0.05** | Reject H₀ → likely stationary             |
| **p-value ≥ 0.05** | Fail to reject H₀ → likely non-stationary |

Python:


from statsmodels.tsa.stattools import adfuller

result = adfuller(series)

print("ADF Statistic:", result[0])
print("p-value:", result[1])

if result[1] < 0.05:
    print("The series is stationary.")
else:
    print("The series is non-stationary.")


3. KPSS Test

The KPSS test can be used as a complementary test.

Here the hypotheses are reversed:

- H₀: Series is stationary.
- H₁: Series is non-stationary.

Using ADF and KPSS together can provide stronger evidence.

How to Transform a Non-Stationary Series

If a series is non-stationary, common transformations include:

1. Differencing

Subtract the previous observation from the current observation:

$$ Y'_t = Y_t - Y_{t-1} $$

Example:

Original:

100, 110, 125, 140, 155

First difference:

10, 15, 15, 15

Differencing is particularly useful for removing trend.

Python:

diff_series = series.diff().dropna()

If necessary, second-order differencing can be applied:

diff2_series = series.diff().diff().dropna()

However, excessive differencing should be avoided.

2. Log Transformation

When the variance increases as the level of the series increases, a logarithmic transformation can help stabilize the variance.

Yt′​=log(Yt​)

import numpy as np

log_series = np.log(series)

This is commonly useful for quantities such as sales, population, or revenue that grow exponentially.

3. Seasonal Differencing

For seasonal patterns, subtract the observation from the same season in the previous cycle:

$$ Y'_t = Y_t - Y_{t-s} $$

where \(s\) is the seasonal period.

For monthly data with yearly seasonality:

$$ Y'_t = Y_t - Y_{t-12} $$

seasonal_diff = series.diff(12).dropna()

Complete Process

Non-stationary series

↓

Plot the series

↓

Perform ADF/KPSS test

↓

Apply transformation

- Trend → Differencing
- Increasing variance → Log transformation
- Seasonality → Seasonal differencing
- Combination → Appropriate combination of transformations

↓

Test again

↓

Stationary series

In short

Stationarity means that the mean, variance, and autocorrelation structure of a time series do not systematically change over time.

A non-stationary series can commonly be converted into a stationary one using differencing, log transformations, and seasonal differencing, followed by another stationarity test. Stationarity is important for models such as ARIMA, because these models generally work best when the input series is stationary.


# 5.  Differentiate between AR, MA, ARIMA, SARIMA, and SARIMAX models in terms of structure and application.

- Difference Between AR, MA, ARIMA, SARIMA, and SARIMAX

These are time-series forecasting models used to model and predict values based on past observations, past errors, trends, and seasonality.

Comparison Table

| Model       | Full Form                                | Main Structure                 | Handles Trend? | Handles Seasonality? | External Variables? | Typical Application                                          |
| ----------- | ---------------------------------------- | ------------------------------ | -------------- | -------------------- | ------------------- | ------------------------------------------------------------ |
| **AR**      | Autoregressive                           | Uses past values of the series | Limited        | No                   | No                  | Predicting stock/temperature values based on previous values |
| **MA**      | Moving Average                           | Uses past forecast errors      | No             | No                   | No                  | Modeling short-term random shocks                            |
| **ARIMA**   | AutoRegressive Integrated Moving Average | AR + Differencing + MA         | Yes            | No                   | No                  | Non-seasonal sales, demand, revenue forecasting              |
| **SARIMA**  | Seasonal ARIMA                           | ARIMA + seasonal components    | Yes            | **Yes**              | No                  | Monthly sales with yearly seasonal patterns                  |
| **SARIMAX** | Seasonal ARIMA with Exogenous Variables  | SARIMA + external variables    | Yes            | **Yes**              | **Yes**             | Sales affected by promotions, price, holidays, weather       |


1. AR – Autoregressive Model

AR predicts the current value using its previous values.

For an AR(2) model:

$$ Y_t = c + \phi_1Y_{t-1}+\phi_2Y_{t-2}+\epsilon_t $$

Where:

- Yt = current value
- Yt−1, Yt−2 = previous values
- ϕ1, ϕ2 = model coefficients
- ϵt = random error

Example: Predict today's electricity demand using demand from the previous few hours.

Key idea:

AR = prediction from past values.

2. MA – Moving Average Model

An MA model predicts the current value using past forecast errors.

For MA(2):

$$ Y_t = c+\epsilon_t+\theta_1\epsilon_{t-1}+\theta_2\epsilon_{t-2} $$

Here, the model considers previous errors to improve the current prediction.

Example: If a forecasting model consistently underestimated demand during the previous few periods, those errors can help improve the current forecast.

Key idea:

MA = prediction from past errors.

3. ARIMA

ARIMA = AutoRegressive Integrated Moving Average

It combines three components:

- AR (p): uses past observations
- I (d): uses differencing to make a non-stationary series stationary
- MA (q): uses past forecast errors

It is represented as:

$$ ARIMA(p,d,q) $$

For example:

ARIMA(1,1,1) means:

- 1 AR term
- 1 order of differencing
- 1 MA term

Application: Forecasting monthly sales where there is a trend but no strong seasonal pattern.

Key idea:

ARIMA = AR + Differencing + MA.

4. SARIMA

SARIMA = Seasonal ARIMA

SARIMA extends ARIMA by adding seasonal AR, differencing, and MA components.

It is represented as:

$$ SARIMA(p,d,q)(P,D,Q)_s $$

Where:

- p,d,q = non-seasonal ARIMA parameters
- P,D,Q = seasonal parameters
- s = seasonal period

For monthly data with yearly seasonality:

$$ s=12 $$
Example

A clothing retailer has sales that:

- increase over several years → trend
- increase every December → seasonality

SARIMA can model both the trend and the repeating yearly pattern.

Key idea:

SARIMA = ARIMA + Seasonality.

5. SARIMAX

SARIMAX = Seasonal ARIMA with Exogenous Variables

SARIMAX extends SARIMA by including external/exogenous variables that influence the time series.

For example, a company's sales may depend on:

- Advertising expenditure
- Product price
- Discounts
- Holidays
- Weather
- Economic indicators

The model can be written conceptually as:

$$ Y_t = \text{SARIMA components} + \beta X_t + \epsilon_t $$

where Xt represents external variables.

Example

A retailer wants to forecast monthly sales using:

- Previous sales
- Seasonal patterns
- Discounts
- Advertising spending
- Holiday periods

SARIMAX is appropriate because it can incorporate these external factors.

Key idea:

SARIMAX = SARIMA + External Variables.

Easy Way to Remember

AR

↓

Past values



MA

↓

Past errors



ARIMA


↓

AR + Differencing + MA



SARIMA

↓

ARIMA + Seasonality



SARIMAX

↓


SARIMA + External Variables


Final Summary
- AR: depends on previous observations.
- MA: depends on previous forecast errors.
- ARIMA: handles non-seasonal trends using differencing.
- SARIMA: extends ARIMA to handle seasonal patterns.
- SARIMAX: extends SARIMA by also using external variables.

Example progression:
If you are forecasting sales → use ARIMA for non-seasonal data → SARIMA when sales have seasonality → SARIMAX when factors such as promotions, prices, or holidays also affect sales.


# 6.  Load a time series dataset (e.g., AirPassengers), plot the original series, and decompose it into trend, seasonality, and residual components ● NYC Taxi Fare Data ● AirPassengers Dataset

-  Python Program: Time Series Decomposition

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose

# Load AirPassengers dataset
# If using a CSV file:
df = pd.read_csv("AirPassengers.csv")

# Convert Month column to datetime
df["Month"] = pd.to_datetime(df["Month"])

# Set Month as the index
df.set_index("Month", inplace=True)

# Create the time series
series = df["#Passengers"]

# ------------------------------------------------
# 1. Plot the original time series
# ------------------------------------------------
plt.figure(figsize=(10, 5))
plt.plot(series)
plt.title("AirPassengers - Original Time Series")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.grid(True)
plt.show()

# ------------------------------------------------
# 2. Decompose the time series
# ------------------------------------------------
decomposition = seasonal_decompose(
    series,
    model="multiplicative",
    period=12
)

# ------------------------------------------------
# 3. Plot decomposition
# ------------------------------------------------
decomposition.plot()
plt.suptitle("AirPassengers Time Series Decomposition", fontsize=14)
plt.tight_layout()
plt.show()
```

1. Original Series

The original AirPassengers series shows an overall increasing trend with a repeating yearly pattern. The passenger numbers rise from 112 thousand in January 1949 to 432 thousand in December 1960.

2. Trend Component

The trend shows the long-term increase in the number of airline passengers.

Interpretation: Air travel increased substantially throughout 1949–1960.

3. Seasonal Component

The seasonal component shows a repeating 12-month pattern, because the data are monthly and passenger numbers tend to be higher during particular months of each year.

4. Residual Component

The residual represents the irregular variation left after removing the trend and seasonal effects.

It can be thought of as:

$$ Residual = \frac{Observed}{Trend \times Seasonal} $$

for a multiplicative decomposition.

Why use model="multiplicative"?

The seasonal fluctuations become larger as the overall number of passengers increases, so a multiplicative decomposition is appropriate for this dataset.



```python
### 7. Apply Isolation Forest on a numerical dataset (e.g., NYC Taxi Fare) to detect anomalies. Visualize the anomalies on a 2D scatter plot.


import numpy as np
import matplotlib.pyplot as plt
from sklearn.ensemble import IsolationForest

# -----------------------------------------
# 1. Create a numerical dataset
# -----------------------------------------
np.random.seed(42)

# Normal data
X_normal = np.random.randn(300, 2)

# Add some abnormal/anomalous points
X_anomalies = np.array([
    [4, 4],
    [5, 5],
    [-4, -4],
    [5, -4],
    [-5, 4]
])

# Combine normal and anomalous data
X = np.vstack([X_normal, X_anomalies])

# -----------------------------------------
# 2. Apply Isolation Forest
# -----------------------------------------
model = IsolationForest(
    contamination=0.05,
    random_state=42
)

model.fit(X)

# Predict anomalies
# 1 = normal, -1 = anomaly
predictions = model.predict(X)

# Identify anomalies
anomalies = predictions == -1
normal = predictions == 1

# -----------------------------------------
# 3. Print results
# -----------------------------------------
print("Total observations:", len(X))
print("Number of normal points:", sum(normal))
print("Number of anomalies:", sum(anomalies))

# -----------------------------------------
# 4. Visualize anomalies
# -----------------------------------------
plt.figure(figsize=(8, 6))

plt.scatter(
    X[normal, 0],
    X[normal, 1],
    label="Normal"
)

plt.scatter(
    X[anomalies, 0],
    X[anomalies, 1],
    marker="X",
    s=100,
    label="Anomaly"
)

plt.title("Isolation Forest - Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()


### How it works

### Isolation Forest isolates observations using randomly selected features and split values.

### - Normal observations are usually in dense regions and require more splits to isolate.
### - Anomalous observations are often isolated quickly.
### - 1 → normal observation
### - -1 → anomaly

### In the scatter plot, the X-shaped points represent detected anomalies.

### Important parameter

### contamination=0.05

### This tells the model that approximately 5% of the observations are expected to be anomalies.
```

# 8. Train a SARIMA model on the monthly airline passengers dataset. Forecast the next 12 months and visualize the results.

- Below is an exam-ready Python program that trains a SARIMA model on the monthly AirPassengers dataset, forecasts the next 12 months, and visualizes the historical data together with the forecast.

```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.statespace.sarimax import SARIMAX

# ---------------------------------------------------
# 1. Load the AirPassengers dataset
# ---------------------------------------------------
df = pd.read_csv("AirPassengers.csv")

# Convert Month column to datetime
df["Month"] = pd.to_datetime(df["Month"])

# Set Month as index
df.set_index("Month", inplace=True)

# Select passenger series
series = df["#Passengers"]

# Make sure the frequency is monthly
series = series.asfreq("MS")

# ---------------------------------------------------
# 2. Train SARIMA model
# ---------------------------------------------------
# SARIMA(p,d,q)(P,D,Q,s)
# Monthly seasonality -> s = 12
model = SARIMAX(
    series,
    order=(1, 1, 1),
    seasonal_order=(1, 1, 1, 12),
    enforce_stationarity=False,
    enforce_invertibility=False
)

model_fit = model.fit(disp=False)

# Display model summary
print(model_fit.summary())

# ---------------------------------------------------
# 3. Forecast the next 12 months
# ---------------------------------------------------
forecast = model_fit.get_forecast(steps=12)

forecast_values = forecast.predicted_mean
confidence_interval = forecast.conf_int()

print("\nForecast for the next 12 months:")
print(forecast_values)

# ---------------------------------------------------
# 4. Visualize actual data and forecast
# ---------------------------------------------------
plt.figure(figsize=(12, 6))

# Historical data
plt.plot(
    series,
    label="Actual Passengers"
)

# Forecast
plt.plot(
    forecast_values,
    label="12-Month Forecast"
)

# Confidence interval
plt.fill_between(
    confidence_interval.index,
    confidence_interval.iloc[:, 0],
    confidence_interval.iloc[:, 1],
    alpha=0.2,
    label="95% Confidence Interval"
)

plt.title("SARIMA Forecast of Airline Passengers")
plt.xlabel("Year")
plt.ylabel("Number of Passengers (thousands)")
plt.legend()
plt.grid(True)
plt.show()
```

The model is:

$$ SARIMA(1,1,1)(1,1,1,12) $$

where:

- p = 1 → one autoregressive term
- d = 1 → first-order differencing
- q = 1 → one moving-average term
- P = 1 → seasonal autoregressive term
- D = 1 → seasonal differencing
- Q = 1 → seasonal moving-average term
- s = 12 → monthly seasonality


The graph will show:

- Actual passenger numbers → historical AirPassengers data
- Forecast → predicted passenger numbers for the next 12 months
- Shaded region → 95% confidence interval

Interpretation

The SARIMA model captures both the trend and 12-month seasonal pattern in airline passenger numbers. The forecast extends the historical series by one year while providing a confidence interval to indicate forecast uncertainty.


# 9.  Apply Local Outlier Factor (LOF) on any numerical dataset to detect anomalies and visualize them using matplotlib.

-  Here is an exam-ready Python program using Local Outlier Factor (LOF) on a synthetic numerical dataset. It detects anomalies and visualizes them using Matplotlib.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import LocalOutlierFactor

# ---------------------------------------------------
# 1. Create a numerical dataset
# ---------------------------------------------------
np.random.seed(42)

# Normal data
X_normal = np.random.randn(300, 2)

# Add some abnormal points
X_anomalies = np.array([
    [4, 4],
    [5, 5],
    [-4, -4],
    [5, -4],
    [-5, 4]
])

# Combine normal data and anomalies
X = np.vstack([X_normal, X_anomalies])

# ---------------------------------------------------
# 2. Apply Local Outlier Factor
# ---------------------------------------------------
lof = LocalOutlierFactor(
    n_neighbors=20,
    contamination=0.05
)

predictions = lof.fit_predict(X)

# LOF returns:
#  1  -> Normal
# -1  -> Anomaly

normal = predictions == 1
anomalies = predictions == -1

# ---------------------------------------------------
# 3. Print results
# ---------------------------------------------------
print("Total observations:", len(X))
print("Number of normal points:", np.sum(normal))
print("Number of anomalies:", np.sum(anomalies))

# ---------------------------------------------------
# 4. Visualize the anomalies
# ---------------------------------------------------
plt.figure(figsize=(8, 6))

# Normal observations
plt.scatter(
    X[normal, 0],
    X[normal, 1],
    label="Normal"
)

# Anomalies
plt.scatter(
    X[anomalies, 0],
    X[anomalies, 1],
    marker="X",
    s=120,
    label="Anomaly"
)

plt.title("Local Outlier Factor (LOF) - Anomaly Detection")
plt.xlabel("Feature 1")
plt.ylabel("Feature 2")
plt.legend()
plt.grid(True)
plt.show()
```

Explanation

Local Outlier Factor (LOF) identifies anomalies by comparing the local density of each point with the density of its neighboring points.

- n_neighbors=20 → each observation is compared with its 20 nearest neighbors.
- contamination=0.05 → approximately 5% of observations are expected to be anomalies.
- fit_predict() returns:
- 1 → normal observation
- -1 → anomaly

Interpretation

In the scatter plot:

- Normal points form the main dense population.
- Points marked with X are identified by LOF as anomalies.
- LOF is particularly useful when an observation is unusual relative to its local neighborhood, making it useful for applications such as fraud detection, network monitoring, and unusual customer behavior.

Recommended approach: Isolation Forest

I would primarily use Isolation Forest because power-grid data can be large and continuously arriving. It is relatively scalable and does not require predefined normal clusters.

For example:

from sklearn.ensemble import IsolationForest

model = IsolationForest(
    n_estimators=100,
    contamination=0.01,
    random_state=42
)

model.fit(training_features)

prediction = model.predict(new_data)

if prediction[0] == -1:
    print("ALERT: Anomalous energy consumption")


Important: I would not rely only on a generic anomaly model. A sudden increase in demand may be completely normal during a hot summer afternoon. Therefore, anomalies should also be assessed relative to time, region, weather, and expected demand.

3. Time-series forecasting

For short-term forecasting, I would choose SARIMAX.

Why SARIMAX?

The data has:

- Strong time dependence
- Daily seasonality — demand patterns repeat throughout the day
- Weekly effects — weekdays and weekends differ
- Weather variables that can influence electricity demand

SARIMAX allows us to model both the time series and external variables:

$$ Y_t = ARIMA\ components + \beta X_t + error_t $$

where Xt could include:

- Temperature
- Humidity
- Weather condition
- Holiday indicator
- Day of week
- Region

For 15-minute data, daily seasonality corresponds to:

$$ 24 \times 4 = 96 $$

observations per day.

So a suitable seasonal model could be conceptually represented as:

$$ SARIMAX(p,d,q)(P,D,Q)_{96} $$

The exact parameters should be selected using validation rather than assumed beforehand.

Comparison

| Model       | Suitable when                                  |
| ----------- | ---------------------------------------------- |
| **ARIMA**   | Non-seasonal time series                       |
| **SARIMA**  | Time series with seasonal patterns             |
| **SARIMAX** | Seasonal demand affected by external variables |


Therefore, SARIMAX is appropriate when weather and other external variables are available and useful.

4. Validation strategy

Because this is time-series data, I would not randomly shuffle the observations into training and testing sets.

Instead, I would use time-based validation.

Example:

Past data                         Future data
|-------------------------|------------------|
       Training                 Validation

     Or rolling/expanding-window validation:

Train → Validate

Train ─────→ Validate

Train ───────────→ Validate

Train ─────────────────→ Validate


Forecasting metrics

I would monitor:

- MAE – average absolute forecasting error
- RMSE – penalizes large forecasting errors
- MAPE – percentage-based error, with care when actual demand is near zero
- Forecast bias – detects systematic over- or under-prediction

For example:

$$ MAE = \frac{1}{n}\sum |Actual-Forecast| $$
Anomaly-detection metrics

If confirmed anomaly labels are available:

- Precision
- Recall
- F1-score
- False-positive rate
- Detection delay

For a power grid, false alarms are important to monitor, because too many unnecessary alerts can cause alert fatigue.

5. Real-time monitoring

After deployment, I would continuously monitor both the data and models.

Data monitoring

Check for:

- Missing 15-minute readings
- Duplicate timestamps
- Sensor failures
- Impossible energy values
- Sudden changes in data distribution
- Communication delays

Model monitoring

Track:

Forecast Accuracy


       ↓

Prediction Error

       ↓

Drift Detection

       ↓

Retraining Trigger

If forecast errors gradually increase, this could indicate concept drift caused by changing consumption patterns, new infrastructure, weather changes, or changes in customer behavior.

The model could then be retrained periodically or when performance crosses a predefined threshold.

6. Business and operational benefits

This system can provide several practical benefits:

1. Grid balancing

Accurate short-term forecasts help operators estimate upcoming electricity demand and balance generation with consumption.

2. Early fault detection

Anomalous spikes or drops can trigger alerts for possible:

- Equipment problems
- Sensor failures
- Unexpected regional demand
- Power-quality issues
- Unusual consumption patterns

3. Cost reduction

Better demand forecasts can help reduce unnecessary generation and improve resource scheduling.

4. Resource planning

Operators can better plan:

- Power generation
- Battery/storage usage
- Renewable-energy integration
- Backup capacity

 Preventive maintenance

Repeated abnormal patterns around particular equipment or regions can help maintenance teams investigate problems before they become larger failures.

Conclusion

For this power-grid problem, I would use Isolation Forest for scalable real-time anomaly detection and SARIMAX for short-term demand forecasting when weather and other external variables are available. I would validate using rolling time-based evaluation, monitor MAE/RMSE and anomaly-detection metrics continuously, and retrain when performance or data distributions change. This combination supports early anomaly detection, accurate demand planning, better grid balancing, reduced operational costs, and more reliable electricity supply.
